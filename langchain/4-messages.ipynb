### Messages

Important object definitions for messages and fundamental unit if context for models in Langchain. They contain role, content and metadata. "role" defines the type(AIMessage, ToolMessage)

##### Text Prompts vs Message Prompts

Text prompts are generally just strings; they ae used for straigtforward generation when retaining convo history is not priority.

Message prompts can be used in lenghty conversations as they handle convo history but they include a bit of code complexity. A list of messages passed as message objects is required.

Message Types: System Message, Human Message, AI Message, Tool Message
